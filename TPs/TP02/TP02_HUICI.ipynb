# **Consignas** 

A partir del dataset Education Data<sup>3</sup> del World Bank Group, desarrollar los flujos necesarios para generar los siguientes archivos de salida (en formato CSV): 

## 1. **CSV de Países con la siguiente estructura:** 

- Campos requeridos: 

   - id: Identificador numérico entero autoincremental (1, 2, 3, ...). 

   - nombre_pais: Nombre oficial del país o entidad regional. 

   - codigo_pais: Código estándar de 3 letras (ISO Alpha-3). 

- Transformaciones esperadas: 

   - Limpieza de registros nulos o datos faltantes en los campos base. 

   - Eliminación de duplicados. 

```python
import pandas as pd
```

```python
df = pd.read_csv("API_4_DS2_en_csv_v2_3434744.csv", encoding="utf-8-sig")
print(df.shape)
display(df.head())
df.info()
```

```python
df_paises = (
    df[["Country Name", "Country Code"]]
    .rename(columns={"Country Name": "nombre_pais", "Country Code": "codigo_pais"})
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
)

df_paises.insert(0, "id", range(1, len(df_paises) + 1))

display(df_paises.head(10))
df_paises.info()
```

```python
df_paises.to_csv("data_ETL/paises.csv", index=False)
```

## **2. CSV de Preguntas con la siguiente estructura:** 

   - Campos requeridos: 

      - id: Identificador numérico entero autoincremental. 

      - pregunta: Descripción corta del indicador o pregunta educativa. 

   - Transformaciones esperadas: 

      - Tratamiento de texto: Expresión regular (RegEx) o manipulación de cadenas para remover completamente cualquier contenido que se encuentre entre paréntesis (incluyendo los propios paréntesis) y aplicar un trim para eliminar espacios extra resultantes. 

      - Consolidar indicadores únicos sin duplicados. 

```python
df_preguntas = (
    df[["Indicator Name"]]
    .rename(columns={"Indicator Name": "pregunta",})
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
)

df_preguntas["pregunta"] = (df_preguntas["pregunta"].str.replace(
    r"\s*\([^)]*\)",
    "",
    regex=True
).str.strip())

df_preguntas = df_preguntas.drop_duplicates().reset_index(drop=True)
df_preguntas = df_preguntas.dropna()

df_preguntas.insert(0, "id", range(1, len(df_preguntas) + 1))

display(df_preguntas.head(20))
```

```python
df_preguntas.to_csv("data_ETL/preguntas.csv", index=False)
```

## **3. CSV de Educación primaria en años ordenado por duración de mayor a menor según el año 2023** 

   - Campos requeridos: 

      - País / Código de País. 

      - Duración de la educación primaria en años. 

   - Transformaciones esperadas: 

      - Filtrar exclusivamente el indicador correspondiente a la duración de la educación primaria en años. 

      - Filtrar o considerar únicamente la medición correspondiente al año 2023. 

      - Tratar valores nulos o no reportados para dicho año (especificar en el informe la estrategia adoptada: imputación, descarte, etc.). 

      - Ordenamiento: Ordenar los datos de mayor a menor según la duración en años. 

```python
df_primaria_total = df[df.iloc[:, 3] == "SE.PRM.DURS"]

df_primaria_2023 = (
    df_primaria_total[["Country Name", "Country Code", "2023"]]
    .rename(columns={"Country Name": "nombre_pais", "Country Code": "codigo_pais", "2023": "dur_prm_anios_2023"})
    .dropna()
    .reset_index(drop=True)
)
df_primaria_2023["dur_prm_anios_2023"] = df_primaria_2023["dur_prm_anios_2023"].astype(int)

df_primaria_2023 = df_primaria_2023.sort_values(["dur_prm_anios_2023", "codigo_pais"], ascending=[False, True]).reset_index(drop=True)

df_primaria_2023.info()
df_primaria_2023.head()
```

```python
df_primaria_2023.to_csv("data_ETL/primaria_2023.csv", index=False)
```
